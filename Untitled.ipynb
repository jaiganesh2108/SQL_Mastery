```python
%%sql
create table CustomerTransactions (
    id int primary key,
    login_device varchar(50),
    customer_name varchar(100),
    ip_address varchar(20),
    product varchar(100),
    amount decimal(10,2),
    is_placed boolean,
    is_viewed boolean,
    transaction_status varchar(20)
);
```

```python
%%sql
select count(*) from CustomerTransactions where transaction_status='completed';
```

```python
%%sql
select count(*) from CustomerTransactions where transaction_status='completed' and product='Laptop';
```

```python
%%sql 
select sum(amount) as total_amount from CustomerTransactions where transaction_status='completed';
```

```python
%%sql
select min(amount) as min_transaction from CustomerTransactions where transaction_status='completed';
```

```python
%%sql
SELECT customer_name, amount
FROM CustomerTransactions
WHERE transaction_status = 'completed'
ORDER BY amount DESC
LIMIT 1;
```

```python
%%sql
SELECT customer_name, amount
FROM CustomerTransactions
WHERE transaction_status = 'completed'
AND amount = (
    SELECT MAX(amount)
    FROM CustomerTransactions
    WHERE transaction_status = 'completed'
);
```

```python
%%sql
select avg(amount) as avg_amount from CustomerTransactions;
```

```python
%%sql
select sum(amount) as total_amount, login_device from CustomerTransactions group by login_device;
```

```python
%%sql
select sum(amount) as total_amount, login_device from CustomerTransactions group by login_device having sum(amount) > 80000;
```

```python
%%sql
select sum(amount) as total_amount, login_device from CustomerTransactions where transaction_status='completed' group by login_device;
```
